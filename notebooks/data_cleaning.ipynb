# 1- Liste des bibliothèques 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
```

# 2- Chargement de la dataset 

```python
# Chemin vers le fichier original
file_path = Path("../data/dataset_original.csv")

# Chargement du dataset
df = pd.read_csv(file_path)

# Afficher la taille du dataset
print("Nombre de lignes :", df.shape[0])
print("Nombre de colonnes :", df.shape[1])

# Afficher les premières lignes
print(df.head())
```

# 3- Analyse des données de la dataset

```python
# Afficher les colonnes du dataset
print(df.columns.tolist())
print (df.info())

# Afficher les statistiques descriptives du dataset
df.describe(include="all").T
```

```python
# Examiner les modalités de chaque variable
for column in df.columns:

    print("\n" + "=" * 60)
    print(f"COLONNE : {column}")
    print("=" * 60)

    # Compter toutes les modalités, y compris les valeurs manquantes
    counts = df[column].value_counts(dropna=False)

    # Afficher toutes les modalités
    print(counts.to_string())

    # Afficher les totaux
    print("\nNombre de modalités différentes :", len(counts))
    print("Total des occurrences :", counts.sum())
    print("Nombre total de lignes du dataset :", len(df))
```

# 4- Nettoyage des données 

## 4-1- Nettoyage des erreurs de saisies  

```python
df["Age"] = df["Age"].astype("string").str.replace(" yrs", "", regex=False).str.strip()
df["AI_Usage_Hours_Per_Day"] = df["AI_Usage_Hours_Per_Day"].astype("string").str.replace(" hrs", "", regex=False).str.strip()
df["Time_Saved_Hours_Per_Week"] = df["Time_Saved_Hours_Per_Week"].astype("string").str.replace(" hrs", "", regex=False).str.strip()
df["Work_or_Study_Hours_Per_Day"] = df["Work_or_Study_Hours_Per_Day"].astype("string").str.replace(" hrs", "", regex=False).str.strip()
```

## 4-3- Gestion des valeurs manquantes

```python
# Compter les valeurs manquantes pour chaque colonne
missing_values = df.isnull().sum()

missing_percentage = (
    df.isnull().mean() * 100
).round(2)

missing_report = pd.DataFrame({
    "Valeurs_manquantes": missing_values,
    "Pourcentage": missing_percentage
})

print(missing_report)


# Remplacer la valeurs manquantes pour les variables qualitatives par une modalité explicite
columns_to_fill = ["Age","Gender", "Country", "Subscription Status", 
                   "Education_Level", "Profession", "AI_Tool", "AI_Purpose", 
                   "Tasks_Performed" ,"Would_Recommend"]

for column in columns_to_fill:
    if column in df.columns:
        df[column] = df[column].fillna("Non renseigné")


```

## 4-2- Gestion des doublons 

```python
# Identifier les doublons dans le dataset
print("Nombre de doublons :", df.duplicated().sum())

# Supprimer les doublons 
df = df.drop_duplicates()

print("Nombre de lignes après suppression :", len(df))
```

# 5- Transformation des données 

## 5-1- Transformation des colonnes 

```python
# Renommer les colonnes pour qu'elles soient plus compréhensibles
column_mapping = {
    "Age": "Age",
    "Gender": "Genre",
    "Country": "Pays",
    "User_Type": "Type d'utilisateur",
    "Education_Level": "Niveau d'éducation",
    "Profession": "Profession",
    "Monthly_Income": "Revenu mensuel",
    "AI_Tool": "IA utilisée",
    "AI_Purpose": "Objectif d'utilisation de l'IA",
    "AI_Usage_Hours_Per_Day": "Utilisation par jour",
    "Tasks_Performed": "Tâches effectuées",
    "Monthly_AI_Cost": "Coût mensuel",
    "Productivity_Score": "Score de productivité",
    "Accuracy_Rating": "Précision perçue",
    "Satisfaction_Score": "Niveau de satisfaction",
    "Time_Saved_Hours_Per_Week": "Gain de temps par semaine",
    "Work_or_Study_Hours_Per_Day": "Heures de travail ou d'étude par jour",
    "Would_Recommend": "Avis sur la recommandation",

}

df = df.rename(columns=column_mapping)

# Vérification
print(df.columns.tolist())
```

## 5-2- Transformation du type de données 

```python
# Type de données 
print(df.dtypes)

# Conversion des variables numériques 
df["Revenu mensuel"] = pd.to_numeric(df["Revenu mensuel"], errors="coerce")
df["Utilisation par jour"] = pd.to_numeric(df["Utilisation par jour"], errors="coerce")
df["Coût mensuel"] = pd.to_numeric(df["Coût mensuel"], errors="coerce")
df["Gain de temps par semaine"] = pd.to_numeric(df["Revenu mensuel"], errors="coerce")
df["Heures de travail ou d'étude par jour"] = pd.to_numeric(df["Heures de travail ou d'étude par jour"], errors="coerce")
print(df.dtypes)
```

## 5-3- Transformation des modalités

```python
# Traduction des modalitées de certaines variables qualitatives pour qu'elles soient plus compréhensibles
value_mapping = {

    "Genre": {
        "Male": "Homme",
        "MALE": "Homme",
        "male": "Homme",
        "M": "Homme",
        "Female": "Femme",
        "female": "Femme",
        "FEMALE": "Femme",
        "F": "Femme",
        "other": "Autre",
        "Other": "Autre",
        "Non-binary": "Non-binaire",
    },

    "Type d'utilisateur": {
        "Student": "Étudiant",
        "Professional": "Professionnel",
    },

    "Avis sur la recommandation": {
        "Yes": "Oui",
        "No": "Non",
        "Maybe": "Peut-être",
    }
}

# # Appliquer les traductions
for column, mapping in value_mapping.items():
    if column in df.columns:
        df[column] = df[column].replace(mapping)


# Vérifier la traduction
for column in value_mapping.keys():
    if column in df.columns:
        print(f"\n--- {column} ---")
        print(df[column].value_counts(dropna=False))
```

## 5-4- Correction des NaN pour les variables qualitatives

```python
# Remplacer la valeurs manquantes pour les variables quantitatives par la médiane
columns_to_fill_median = ["Revenu mensuel", "Utilisation par jour", "Coût mensuel", "Score de productivité",
                          "Précision perçue", "Niveau de satisfaction", "Gain de temps par semaine", "Heures de travail ou d'étude par jour"]
                          
for col in columns_to_fill_median:
    df[col] = df[col].fillna(df[col].median())
```

# 6- Vérification de la cohérance des données 

```python
# Afficher les modalités uniques pour chaque variable qualitative
for column in df.select_dtypes(include="object").columns:
    print(f"\n{column}")
    print(df[column].unique())


# Vérifier les valeurs manquantes et les doublons après le nettoyage
print("Dimensions finales :", df.shape)

print("\nValeurs manquantes :")
print(df.isnull().sum())

print("\nDoublons :")
print(df.duplicated().sum())

print("\nTypes des colonnes :")
print(df.dtypes)
```

# 7- Export de la dataset nettoyée  

```python
# Export de la dataset en format CSV
from pathlib import Path

output_dir = Path("../data")
output_dir.mkdir(parents=True, exist_ok=True)


df.to_csv(output_dir / "dataset_clean.csv", index=False,  encoding="utf-8-sig")

print("Fichier CSV exporté avec succès !")


# Export de la dataset en format JSON
df.to_json(output_dir / "dataset_clean.json", orient="records", force_ascii=False, indent=4)

print("Fichier JSON exporté avec succès !")
```
